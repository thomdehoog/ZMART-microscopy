# Example workflow for coders

Drive a microscope with the ZMART Controller.

See which microscopes are installed.

```python
from zmart_controller import mic

mic.get_instruments()
```

Connect to one of them.

```python
mic.connect("stellaris")
```

Where is the stage?

```python
mic.get_xyz()
```

Take a picture.

```python
mic.acquire("A1", acquisition_type="overview", output_folder="C:/zmart-runs/example")
```

Disconnect.

```python
mic.disconnect()
```
