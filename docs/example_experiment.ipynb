# A first experiment

An overview scan, then a detailed scan at a few positions. The notebook talks
only to the controller, so it runs on any microscope with a driver.

1. Connect
2. Capture two settings
3. Choose the positions
4. Acquire
5. Disconnect

## Setup

Plug in the mock microscope, a driver like any other. It lives in memory, so
nothing else is needed. On a real microscope the driver is already plugged in
and this cell goes away.

```python
import zmart_controller

zmart_controller.register_driver("zmart_driver_mock", remember=False)
```

## 1. Connect

`get_instruments()` lists the microscopes on this computer. Pick one by name
and connect.

```python
import zmart_controller

zmart_controller.get_instruments()
```

```python
instrument = next(i for i in zmart_controller.get_instruments() if i["microscope"] == "mock-scope")
zmart_controller.set_instrument(instrument)
```

## 2. Capture two settings

`get_state()` captures the instrument's settings. Every command answers with
`success` and `report`; the state is the report. Its `changeable` part is what
`set_state()` applies later. We capture one gentle and one detailed setting.

```python
overview = zmart_controller.get_state()["report"]
overview["changeable"]["laser_power"] = 2.0
overview
```

```python
detail = zmart_controller.get_state()["report"]
detail["changeable"]["laser_power"] = 8.0
detail["changeable"]["gain"] = 2.0
detail
```

## 3. Choose the positions

Positions are micrometers from the microscope's origin. `get_xyz()` shows the
current position and how far each axis can travel.

```python
zmart_controller.get_xyz()["report"]
```

```python
positions = [
    {"x": 0.0, "y": 0.0, "z": 0.0},
    {"x": 120.0, "y": 0.0, "z": 0.0},
    {"x": 0.0, "y": 120.0, "z": 0.0},
]
```

## 4. Acquire

Apply a setting, then visit each position and capture. `acquire()` captures
and saves in one call. The report says where the files went. Run it again with
`detail` for the detailed scan.

```python
zmart_controller.set_state(overview)
```

```python
for i, pos in enumerate(positions):
    zmart_controller.set_xyz(pos["x"], pos["y"], pos["z"])
    answer = zmart_controller.acquire(acquisition_type="overview", position_label=f"pos_{i:03d}")
    print(answer["success"], answer["report"])
```

## 5. Disconnect

```python
zmart_controller.disconnect()
```
